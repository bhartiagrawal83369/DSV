```python
#vectorized computation = vectorized means applying an operation to an entire array without writing an explicit python loop
```

```python

#traditional python approach
numbers = [1,2,3,4,5]
result = []
for x in numbers:
   result.append(x*2)
```

```python
#numpy vectorized approach
arr = np.array([1,2,3,4,5])
result = arr*2
print(result)
```

```python
#universal function
#unary universal function
arr = np.array([1,4,9,16])
print(np.sqrt(arr))
```

```python
arr = np.array([1,4,9,16])
print(np.square(arr))
```

```python
#binary universal function
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.add(a,b))
```

```python
print(np.subtract(a,b))
```

```python
print(np.multiply(a,b))
```

```python
print(np.divide(a,b))
```

```python
print(np.maximum(a,b))
```

```python
#array operations
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** 2)
```

```python
#scalar operations
arr = np.array([1,2,3,4])
print(arr * 10)
```

```python
#comparison and boolean indexing
arr = np.array([10,20,30,40])
print(arr > 20)
```

```python
#boolean indexing selects values that satisfy a condition
marks = np.array([45,70,85,30,90])
passed = marks[marks >= 40]
print(passed)
```

```python
#aggeration operations
arr = np.array([10,20,30,40,50])
print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
```

```python
#operation along an axis
arr = np.array([[10,20,30],[40,50,60]])
```

```python
#for a 2D array
np.sum(arr)
np.sum(arr, axis=0)
np.sum(arr, axis=1)
```

```python
#reshaping flattering and transpose
```

```python
#reshaping
arr = np.arange(1,13)
new_arr = arr.reshape(3,4)
print(new_arr)
```

```python
#flattening
arr = np.array([[1,2],[3,4]])
print(arr.flatten())
```

```python
#transpose
arr = np.array([[1,2,3],[4,5,6]])
print(arr.T)
```

```python
#Broadcasting
arr = np.array([[1,2,3],[4,5,6]])
print(arr + 10)
```
