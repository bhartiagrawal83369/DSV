```python
#Example 1: Student marks by department and semester
import pandas as pd

data = {
    "Department": ["CSE", "CSE", "CSE", "DS", "DS", "DS"],
    "Semester": [1, 2, 3, 1, 2, 3],
    "Average_Marks":[76, 81, 79, 82, 85, 88]
}
df = pd.DataFrame(data)

multi = df.set_index(["Department", "Semester"])

print(multi)
```

```python
#Example 2: Selecting data from a multiindex
#select all semester records for CSE
print(multi.loc["CSE"])

#Select the average marks for CSE , Semester 2
print(multi.loc[("CSE", 2), "Average_Marks"])

#Convert index levels back into ordinary columns
print(multi.reset_index())

```

```python
##3.1 merge(): Match rows using a common key
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Asha", "Ravi", "Neha", "Imran"],
    "Department": ["CSE", "DS", "CSE", "DS"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105],
    "Data_Science_Marks": [86, 91, 78, 88]
})

#keep only present in both datasets
dat1 = pd.merge(students, marks, on="Student_ID", how="inner")
print(dat1)

#keep every student; unmatched marks become NaN
left = pd.merge(students, marks, on="Student_ID", how="left")
print(left)

#keep all IDs from both datasets
outer = pd.merge(students, marks, on="Student_ID", how="outer")
print(outer)
```

```python
## 3.2concat(): Stack datasets
jan = pd.DataFrame({
    "Month": ["Jan", "Jan"],
    "Product": ["Pen", "Notebook"],
    "Units": [150, 95]
})

feb = pd.DataFrame({
    "Month": ["feb", "feb"],
    "Product": ["Pen", "Notebook"],
    "Units": [150, 95]
})

sales = pd.concat([jan, feb], ignore_index=True)
print(sales)
```

```python
##3.3 join(): Combine using index labels
student_info = students.set_index("Student_ID")

student_marks = marks.set_index("Student_ID")

joined = student_info.join(student_marks, how="left")

print(joined)
```

```python
##4.1 pivot(): long format to wide format
sales_long = pd.DataFrame({
    "Product": ["Pen", "Pen", "Notebook", "Notebook"],
    "Month": ["jan", "feb", "jan", "feb"],
    "Units_Sold": [120,150,80,95]
})

sales_wide = sales_long.pivot(
    index="Product",
    columns="Month",
    values="Units_Sold"
)

print(sales_wide)
```
